```python
import re
text = 'a56456gjghj@edu.corp.ru'
re.match(r'\w+@\w+\.\w+', text)
```

```python
text = '+7(999)-999-9999'
re.match(r'\+\d-\d{3}-\d{3}-\d{2}-\d{2}', text)
```

```python
text = 'Голубничий А. А.'
re.match(r'\w+ \w\. \w\.', text)
```

```python
text = 'Новосибирск Ленина 90'
re.match(r'\w+ \w+ \d+', text)
```

1. Шаблон проверки электронной почты:
re.match(r'\w+@\w+\.\w+', text)
2. Шаблон для проверки телефонных номеров
re.match(r'\+\d-\d{3}-\d{3}-\d{2}-\d{2}', text)
3. Шаблон для проверки ФИО
re.match(r'w+ \w\. \w\.')
4. Шаблон для проверки адреса
r'\w+ \w+ \d+'
